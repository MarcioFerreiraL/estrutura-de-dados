1. Ler uma matriz 5x5 e gerar outra em que cada elemento é o cubo do elemento respectivo na matriz original. Imprima depois o elemento do meio dessa nova matriz

```python
# Criando uma matriz 5x5 com números inteiros aleatórios de 1 a 10
matriz_original = np.random.randint(1, 11, size=(5, 5))

# Gerando a nova matriz onde cada elemento é o cubo do original
matriz_cubo = matriz_original ** 3

# Imprimindo a matriz original
print("Matriz Original:")
print(matriz_original)

# Imprimindo a matriz com os cubos
print("\nMatriz Cubo:")
print(matriz_cubo)

# O elemento do meio em uma matriz 5x5 está no índice (2, 2)
elemento_meio = matriz_cubo[2, 2]
print(f"\nO elemento do meio da nova matriz é: {elemento_meio}")
```

2. Faça um algoritmo para ler uma matriz 3x3 real e imprimir a soma dos elementos da
diagonal principal. Generalize para uma matriz NxN (2x2, 3x3, 4x4... NxN).

```python

try:
    n = int(input("Digite a dimensão N para a matriz NxN: "))

    # Criando uma matriz NxN com números aleatórios
    matriz_nxn = np.random.randint(1, 100, size=(n, n))

    # Calculando a soma da diagonal principal
    soma_diagonal = 0
    for i in range(n):
        soma_diagonal += matriz_nxn[i, i]

    print(f"\nMatriz Gerada: ({n}x{n})")
    print(matriz_nxn)
    print(f"\nA soma da diagonal principal é: {soma_diagonal}")

except ValueError:
    print("Por favor, digite um número inteiro válido.")
```

4. Faça um algoritmo para criar e exibir a matriz identidade 4x4. Não é permitido que o usuário insira valores, a matriz deverá ser construída automaticamente através de laços.

Ex:

Screenshot_1.png

```python
# Criando uma matriz 4x4 vazia (preenchida com zeros)
identidade = np.zeros((4, 4), dtype=int)

# Preenchendo a matriz
for i in range(4):
    for j in range(4):
        if i == j:
            identidade[i, j] = 1
        else:
            identidade[i, j] = 0

print("Matriz Identidade:")
print(identidade)
```

5. Crie uma matriz 7x8 onde cada elemento é a soma dos índices da sua posição dentro da
matriz.


```python
# Definindo as dimensões
linhas= 7
colunas = 8
matriz_soma_indices = np.zeros((linhas, colunas), dtype=int)

# Preenchendo a matriz com a soma dos índices
for i in range(linhas):
    for j in range(colunas):
        matriz_soma_indices[i, j] = i + j

print(f"Matriz {linhas}x{colunas} com a soma dos índices:")
print(matriz_soma_indices)
```

6. Faça um algoritmo para ler duas matrizes reais A e B 5x5, e criar uma matriz C de modo
que: Cij = 1 se Aij da matriz A existe em algum lugar na matriz B; Cij = 0 se Aij da matriz A
não existir em B.


```python
# Criando matrizes A e B 5x5 com números aleatórios
A = np.random.randint(1, 10, size=(5, 5))
B = np.random.randint(1, 10, size=(5, 5))

# Criando a matriz C 5x5, preenchida de zeros
C = np.zeros((5, 5), dtype=int)

# Verificando cada elemento de A
for i in range(5):
    for j in range(5):
        elemento_A = A[i, j]
        if elemento_A in B:
            C[i, j] = 1
        else:
            C[i, j] = 0

print("Matriz A:")
print(A)
print("\nMatriz B:")
print(B)
print("\nMatriz C:")
print(C)
```

7. Faça um algoritmo para ler um vetor com 20 elementos inteiros. Modifique o vetor de
modo que o primeiro elemento passe para a última posição, e desloque todos os outros
elementos uma posição para a esquerda. Imprima depois o vetor.
Ex:
image.png

```python
# Criando um vetor com 20 elementos
vetor = np.arange(1, 21)

print("Vetor Original")
print(vetor)

# Guardando o primeiro elemento
primeiro_elemento = vetor[0]

# Deslocando todos os elementos uma posição para a esquerda
for i in range(len(vetor) - 1):
    vetor[i] = vetor[i + 1]

# Colocando o primeiro elemento na última posição
vetor[-1] = primeiro_elemento

print("\nVetor Modificado:")
print(vetor)
```

8. Uma matriz é triangular superior se todos os elementos abaixo da diagonal principal são 0. Ler uma matriz 4x4 e exibir uma mensagem dizendo se ela é ou não é triangular
superior.

```python
def is_triangular_superior(matriz):
    # Verifica se uma matriz é triangular superior.
    # A matriz precisa ser quadrada
    if matriz.shape[0] != matriz.shape[1]:
        return False

    n = matriz.shape[0]
    # Percorre todos os elementos abaixo da diagonal principal
    for i in range(n):
        for j in range(n):
            if i > j and matriz[i, j] != 0:
                return False
    return True

# Exemplo 1: Matriz que É triangular superior
matriz_ex_1 = np.array([
    [5, 2, 3, 4],
    [0, 8, 6, 7],
    [0, 0, 1, 9],
    [0, 0, 0, 3]
])

# Exemplo 2: Matriz que NÃO É triangular superior
matriz_ex_2 = np.array([
    [5, 2, 3, 4],
    [1, 8, 6, 7],
    [0, 0, 1, 9],
    [0, 0, 0, 3]
])

print("Verificação da Matriz 1:")
print(matriz_ex_1)
if is_triangular_superior(matriz_ex_1):
    print("Resultado: A matriz É triangular superior.")
else:
    print("Resultado: A matriz NÃO É triangular superior.")

print("\nVerificação da Matriz 2:")
print(matriz_ex_2)
if is_triangular_superior(matriz_ex_2):
    print("Resultado: A matriz É triangular superior.")
else:
    print("Resultado: A matriz NÃO É triangular superior.")
```

9. A empresa Pregotex Corporation possui uma tabela para representar as vendas dos seus
produtos (em barris) durante o ano. Cada linha representa um produto e as colunas são
os meses do ano. Veja o exemplo:

image.png

Faça um algoritmo que leia uma tabela deste tipo e depois escreva:

a) O total de venda anual de cada produto;

b) O total de venda da empresa em cada mês;

```python
# Dados da tabela fornecida
dados_vendas = np.array([
    [1200, 1250, 1300, 1300, 1310, 1280, 1290, 1310, 1340, 1345, 1300, 1280], # Prego
    [3100, 3150, 3150, 3200, 3230, 3240, 3200, 3399, 3340, 3345, 3280, 3250], # Porca
    [2500, 2640, 2630, 2700, 2710, 2690, 2670, 2690, 2700, 2710, 2650, 2590], # Arruela
    [3000, 2980, 2995, 3004, 3000, 3100, 3150, 3200, 3250, 3230, 3140, 2999]  # Parafuso
])

produtos = ["Prego", "Porca", "Arruela", "Parafuso"]
meses = ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun", "Jul", "Ago", "Set", "Out", "Nov", "Dez"]

# a) Total de venda anual de cada produto (soma das linhas)
vendas_por_produto = np.sum(dados_vendas, axis=1)

print("Total de Venda Anual por Produto:")
for i in range(len(produtos)):
    print(f"{produtos[i]}: {vendas_por_produto[i]} barris")

# b) Total de venda da empresa em cada mês (soma das colunas)
vendas_por_mes = np.sum(dados_vendas, axis=0)

print("\nTotal de Venda da Empresa por Mês:")
for i in range(len(meses)):
    print(f"{meses[i]}: {vendas_por_mes[i]} barris")
```
